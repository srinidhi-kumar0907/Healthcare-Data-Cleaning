```python
# 2. Check missing values
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# 3. Check and remove duplicates
```

```python
print("Duplicate rows before cleaning:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("Duplicate rows after cleaning:", df.duplicated().sum())
print("New shape:", df.shape)
```

```python
# 4. Correct date data types
```

```python
print(df.dtypes)
```

```python
df["Date of Admission"] = pd.to_datetime(
    df["Date of Admission"],
    errors="coerce"
)

df["Discharge Date"] = pd.to_datetime(
    df["Discharge Date"],
    errors="coerce"
)
print(df[["Date of Admission", "Discharge Date"]].dtypes)
```

```python
# 5. Handle outliers
```

```python
Q1 = df["Billing Amount"].quantile(0.25)
Q3 = df["Billing Amount"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[
    (df["Billing Amount"] < lower_bound) |
    (df["Billing Amount"] > upper_bound)
]
print("Number of outliers:", len(outliers))
```

```python
df["Billing Amount"] = df["Billing Amount"].clip(
    lower=lower_bound,
    upper=upper_bound
)

print("Outliers handled successfully.")
```

```python
# 6. Feature engineering
```

```python
df["Admission Year"] = df["Date of Admission"].dt.year
df["Admission Month"] = df["Date of Admission"].dt.month

df["Length of Stay"] = (
    df["Discharge Date"] - df["Date of Admission"]
).dt.days

df[
    [
        "Date of Admission",
        "Discharge Date",
        "Admission Year",
        "Admission Month",
        "Length of Stay"
    ]
].head()
```

```python
## 7. Profit Margin

Profit margin calculation is not applicable because the provided
healthcare dataset does not contain separate Revenue and Cost columns.
Therefore, a valid profit margin cannot be calculated without
introducing fabricated data.
```

```python
# 8. Standardize text
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()

df["Name"] = df["Name"].str.title()
df["Doctor"] = df["Doctor"].str.title()

print("Text standardization completed.")
```

```python
# 9 .Final check
```

```python
print("Final shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
print("Total duplicate rows:", df.duplicated().sum())
```

```python
# 10. Exporting file
```

```python
df.to_csv("clean_dataset.csv", index=False)

print("clean_dataset.csv created successfully!")
```
